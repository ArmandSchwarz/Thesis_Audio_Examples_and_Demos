```python
from scipy.io.wavfile import read, write
import numpy as np
```

```python
titre = [f'son_consigne_slide_{i}_n{j}.wav' for i in [3,4] for j in [1,2,3]]
```

```python
titre = [f'son_consigne_slide_3_n2.wav']
```

```python
for i in titre:
    fs, data = read(i)
    data = data / np.max(np.abs(data))  # Normalisation
    data = data.astype(np.float32)
    write(i, fs, data)
```

```python
fe, x = read('a_casser_en_2.wav')
x1 = x[:len(x)//2]
x2 = x[len(x)//2:]
x1 = x1 / np.max(np.abs(x1))  # Normalisation
x1 = x1.astype(np.float32)
x2 = x2 / np.max(np.abs(x2))  # Normalisation
x2 = x2.astype(np.float32)
write('son_consigne_slide_2_n1.wav', fe, x1)
write('son_consigne_slide_2_n2.wav', fe, x2)
```
